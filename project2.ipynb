```python
# Project - Student Marks Analyzer

import numpy as np


def students_marks(students):
  num=[]
  for i in range(students):
    x=int(input(f"Enter the marks for students{i+1}: "))
    num.append(x)
  return np.array(num)


def students_stat(arr):
  sum=np.sum(arr) # get sum of all marks
  avg=np.mean(arr) # get average of marks
  max=np.max(arr)  # get highest marks
  min=np.min(arr)  # get minimum marks
  return sum,avg,max,min

def students_grade(arr,students): # checking Grade
  for i in range(students):
    if arr[i]>=85:
      print(f"Grade of student {i+1} is A")
    elif arr[i]>=70 and arr[i]<85:
      print(f"Grade of student {i+1} is B")
    elif arr[i]>=50 and arr[i]<70:
      print(f"Grade of student {i+1} is C")

    else:
      print(f"student {i+1} is  Fail! ")



def display_result(sum,avg,max,min):
  print(f"Sum of all marks is given as: {sum}  ")
  print(f"Average marks is given as: {avg} ")
  print(f"Highest marks is given as: {max} ")
  print(f"Lowest marks is given as: {min} ")


students=int(input("Enter number of students: "))
arr=students_marks(students)

students_grade(arr,students)
um,avg,max,min=students_stat(arr)
display_result(um,avg,max,min)
```
